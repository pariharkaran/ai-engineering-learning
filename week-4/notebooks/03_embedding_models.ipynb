```python
import os

from dotenv import load_dotenv

load_dotenv()
```

```python
from pathlib import Path

from langchain_community.document_loaders import TextLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter

PDF_PATH = Path("../documents/hr_policy.txt")

loader = TextLoader(str(PDF_PATH))
documents = loader.load()

text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=500,
    chunk_overlap=50,
)

chunks = text_splitter.split_documents(documents)

sample_chunks = chunks[:5]

sample_texts = [
    chunk.page_content
    for chunk in sample_chunks
]

print(sample_texts)

# sample_texts = [
#     chunk.page_content
#     for chunk in sample_chunks
# ]
```

```python
from langchain_mistralai import MistralAIEmbeddings

embeddings = MistralAIEmbeddings(
    model="mistral-embed"
)

text = "Employees are entitled to 18 annual paid leaves."

sample_vectors = embeddings.embed_documents(sample_texts)

print(f"Chunks: {len(sample_texts)}")
print(f"Vectors: {len(sample_vectors)}")

for i, vector in enumerate(sample_vectors):
    print(f"Chunk {i}: vector dimensions = {len(vector)}")
```

```python
query = "How many vacation days do employees receive?"

query_vector = embeddings.embed_query(query)
```

## Observations

- Embeddings convert text into numerical vectors.
- Each document chunk receives its own embedding.
- `embed_documents()` is used to embed document chunks.
- `embed_query()` is used to embed a user's query.
- Embedding vectors allow semantic relationships between text to be measured.
- Embedding models are used for retrieval, not answer generation.
- The embedding model and generation LLM can be different models.
- In our RAG pipeline, Mistral embeddings will be used for both document
  chunks and user queries.
- The generated vectors will later be stored in FAISS for similarity search.
