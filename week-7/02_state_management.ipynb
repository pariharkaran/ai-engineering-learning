# Week 7 — State Management

## Learning Goals

In this notebook we will learn:

- What is State?
- How state flows between nodes
- How nodes read state
- How nodes update state
- How multiple nodes share the same state

```python
from typing import TypedDict

from langgraph.graph import StateGraph, START, END

class State(TypedDict):
    name: str
    message: str

initial_state: State = {
    "name": "Karan",
    "message": ""
}

initial_state
```

```python
def name_node(state: State):
    print("Name node received:")
    print(state)

    return {
        "message": f"Hello {state['name']}!"
    }

result = name_node(initial_state)

result
```

```python
def message_node(state: State):
    print("Message node received:")
    print(state)

    return {
        "message": state["message"] + " Welcome to Week 7!"
    }
```

```python
graph = StateGraph(State)
graph.add_node("name_node", name_node)
graph.add_node("message_node", message_node)

graph.add_edge(START, "name_node")
graph.add_edge("name_node", "message_node")
graph.add_edge("message_node", END)

app = graph.compile()

result = app.invoke({
    "name": "Karan",
    "message": ""
})

result
```
