```python
from langchain_core.tools import tool

@tool
def add_numbers(a: int, b: int) -> int:
    """Add two numbers together."""
    return a + b

print(type(add_numbers))
print(add_numbers.name)
print(add_numbers.description)
print(add_numbers.args_schema)

result = add_numbers.invoke({
    "a": 10,
    "b": 20
})

print(result)
```

```python
import os
from dotenv import load_dotenv
from langchain_tavily import TavilySearch

load_dotenv()

search_tool = TavilySearch(max_results=3)

print(type(search_tool))
print(search_tool.name)
print(search_tool.description)
print(search_tool.args)
```

```python
result = search_tool.invoke({
    "query": "What is uranium and where it found on earth?"
})

print(result)
```

```python
from langchain_groq import ChatGroq

llm = ChatGroq(
    model=os.environ.get("GROQ_MODEL_OPEN"),
    temperature=0
)

llm_with_tools = llm.bind_tools([search_tool])

question_1 = "What is the latest stable version of React Native?"
question_2 = "What is 10 + 20?" 

response = llm_with_tools.invoke(question_1)

print(f"LLM Answer: {response.content}")
print(f"response conntent: {response.tool_calls}")
```

```python

response2 = llm_with_tools.invoke(question_2)

print(f"LLM Answer: {response2.content}")
print(f"response conntent: {response2.tool_calls}")
```
